# Pelabelan (Advanced Multi-Level Scoring System)

## 📋 Sistem Pelabelan yang Ditingkatkan

Sistem pelabelan baru menggunakan **Advanced Multi-Level Scoring System** dengan beberapa peningkatan:

### 🎯 Fitur Utama:

1. **Kategori Kata Bertingkat (5 Level)**
   - **Extreme Toxic** (bobot 5): Ancaman, kata kasar seksual
   - **Personal Attack** (bobot 4): Penghinaan personal, diskriminasi
   - **Gaming Toxic** (bobot 3): Toxic spesifik gaming (noob, feeder, dll)
   - **Negative Words** (bobot 1-2): Kata negatif biasa (bukan cyberbullying)

2. **Pattern Detection**
   - Deteksi frasa/pola cyberbullying (misal: "tolol afk", "push lah goblok")
   - Bonus score untuk pola yang terdeteksi

3. **Severity Levels**
   - **Safe**: Score 0
   - **Minor Negative**: Score 1-2
   - **Moderate Toxic**: Score 3-5
   - **Severe Toxic**: Score 6-8
   - **Extreme Cyberbullying**: Score 9+

4. **Smart Categorization**
   - Tidak hanya berdasarkan score, tapi juga jenis kata yang terdeteksi
   - Cyberbullying harus punya score ≥4 DAN kata dari kategori extreme/personal attack
   - Atau score sangat tinggi (≥8) otomatis cyberbullying

### ✅ Perbedaan dengan Sistem Lama:

| Aspek | Sistem Lama | Sistem Baru |
|-------|-------------|-------------|
| Metode | Binary (ada/tidak ada kata toxic) | Scoring dengan bobot bertingkat |
| Akurasi | Rendah (semua kata toxic = cyberbullying) | Tinggi (membedakan negatif vs cyberbullying) |
| Granularitas | 2 kategori | 5 severity levels |
| Pattern Detection | Tidak ada | Ada (frasa/pola) |
| Konteks | Diabaikan | Dipertimbangkan (akumulasi kata, jenis kata) |

### 🔍 Contoh Pembedaan:

- **"anjir parah banget"** → Score 2 → **Non-cyberbullying** (hanya kata negatif ringan)
- **"dasar noob tolol"** → Score 7 → **Cyberbullying** (penghinaan personal + gaming toxic)
- **"bodoh goblok sampah"** → Score 12 → **Cyberbullying** (akumulasi penghinaan)

### 📊 Output yang Dihasilkan:

Setiap komentar akan memiliki:
- **Toxicity Score**: Skor numerik toxicity
- **Severity Level**: Tingkat keparahan
- **Category**: cyberbullying / non-cyberbullying
- **Matched Words**: Kata-kata toxic yang terdeteksi beserta bobotnya

---

**Catatan Penting**: 
- Kata negatif ≠ cyberbullying
- Sistem ini memberikan pelabelan yang lebih akurat untuk training model ML/DL
- Data berlabel bisa langsung digunakan untuk supervised learning

````python
# --- IMPROVED LEXICON-BASED DETECTION V2.0 ---
print("=" * 80)
print("🚀 INITIALIZING IMPROVED LEXICON SYSTEM V2.0")
print("=" * 80)

import re
from collections import Counter

# ============================================================================
# 1. EXPANDED LEXICON (300+ words)
# ============================================================================

# Kategori 1: Extreme Toxic (BOBOT 5) - EXPANDED
extreme_toxic_v2 = {
    # Ancaman & perintah menyakiti diri
    'mati': 5, 'bunuh': 5, 'bunuh diri': 5, 'pergi mati': 5, 'die': 5, 
    'drop dead': 5, 'mati aja': 5, 'mati kamu': 5, 'mati kau': 5,
    'mampus': 5, 'gebuk': 5, 'hajar': 5, 'siksa': 5,
    
    # Kata kasar seksual
    'kontol': 5, 'kntl': 5, 'kintil': 5, 'memek': 5, 'mek': 5, 'pepek': 5,
    'pep3k': 5, 'pepeq': 5, 'peler': 5, 'titit': 5, 'titid': 5, 'tititmu': 5,
    'ngentot': 5, 'entot': 5, 'coli': 5, 'colmek': 5, 'jancok': 5, 'jancuk': 5,
    
    # Kata kasar ekstrem
    'bajingan': 5, 'bangsat': 5, 'bangke': 5, 'brengsek': 5, 'sialan': 5,
    'sial': 4, 'biadab': 5, 'kunyuk': 5, 'kampang': 5,
    'fuck': 5, 'fck': 5, 'fcking': 5, 'fucking': 5, 'fk u': 5, 'motherfucker': 5,
    'bitch': 5, 'bicth': 5, 'btch': 5, 'asshole': 5, 'ashole': 5,
    'tai anjing': 5, 'tai babi': 5, 'tai kucing': 5,
}

# Kategori 2: Personal Attack (BOBOT 4) - EXPANDED
personal_attack_v2 = {
    # Penghinaan intelektual
    'bodoh': 4, 'tolol': 4, 'bego': 4, 'idiot': 4, 'dungu': 4, 'goblok': 4,
    'bodo': 4, 'bodoamat': 4, 'tolil': 4, 'tololll': 4, 'begok': 4, 'begoo': 4,
    'gblk': 4, 'gblg': 4, 'stupid': 4, 'dumb': 4, 'dumbass': 4, 'retard': 4,
    'moron': 4, 'otak udang': 4, 'otak atik': 4, 'otak dongo': 4,
    'bego lu': 4, 'tolol banget': 4, 'goblok banget': 4,
    
    # Variasi L33t speak
    'b0d0h': 4, 't0l0l': 4, 'g0bl0k': 4, 'b3g0': 4,
    
    # Penghinaan fisik/penampilan  
    'jelek': 4, 'burik': 4, 'buruk': 4, 'sampah': 4, 'ampas': 4,
    'buruk rupa': 4, 'jelek banget': 4, 'cacat': 4, 'kere': 4,
    
    # Hewan sebagai penghinaan
    'monyet': 4, 'anjing': 4, 'babi': 4, 'asu': 4, 'kampret': 4,
    'kodok': 4, 'cacing': 4, 'kecebong': 4, 'dog': 4, 'lonte': 4,
    'pelacur': 4, 'perek': 4, 'sundal': 4,
    
    # Variasi anjing
    'anjg': 4, 'anjim': 4, 'njing': 4, 'anying': 4, 'anj1ng': 4,
    
    # Diskriminatif
    'autis': 4, 'hina': 4, 'najis': 4, 'sampah masyarakat': 5,
    'ga berguna': 4, 'tidak berguna': 4, 'mending mati': 5,
}

# Kategori 3: Gaming Toxic (BOBOT 3) - EXPANDED
gaming_toxic_v2 = {
    # Skill-shaming
    'noob': 3, 'nob': 3, 'nub': 3, 'newbie': 3, 'cupu': 3, 'cupuu': 3,
    'lemah': 3, 'lemot': 3, 'lemottt': 3, 'useless': 3, 'ga guna': 3,
    'nub banget': 4, 'noob banget': 4, 'cupu abis': 4,
    
    # Perilaku negatif game
    'feeder': 3, 'feder': 3, 'fdr': 3, 'ngefeed': 3, 'inting': 3,
    'afk': 3, 'afkers': 3, 'beacon': 3, 'beakun': 3, 'troll': 3,
    'savage hunter': 2, 'kill steal': 3, 'ks': 3,
    
    # Toxic player behavior
    'toxic': 3, 'toxik': 3, 'toksek': 3, 'toxic player': 4,
    'player sampah': 4, 'ga ngerti main': 4, 'gaptek': 3,
    
    # Gaming slang
    'sok jago': 3, 'jago kandang': 3, 'alay': 3, 'caper': 2,
    'loser': 3, 'ez game': 3, 'ez clap': 3, 'dogshit': 4, 
    'dog player': 4, 'trash player': 4, 'cancer': 4,
}

# Kategori 4: Negative Words (BOBOT 1-2) - EXPANDED
negative_words_v2 = {
    # Kritik umum
    'parah': 1, 'buruk': 1, 'jelek': 1, 'payah': 2, 'lemah': 2,
    
    # Ekspresi negatif ringan
    'bacot': 2, 'bacod': 2, 'bacott': 2, 'cocot': 2, 'cocod': 2,
    'tai': 2, 'taii': 2, 'tayi': 2, 'taik': 2,
    'shit': 2, 'sh1t': 2, 'shet': 2, 'sheet': 2, 'shitty': 2,
    'goblog': 2, 'goblok': 2,
    
    # Slang negatif
    'anjir': 1, 'anjirr': 1, 'anjay': 1, 'wtf': 2, 'wth': 1,
    'darat': 1, 'garing': 1, 'cringe': 1,
}

# ============================================================================
# 2. NEGATION WORDS (untuk deteksi negasi)
# ============================================================================

NEGATION_WORDS = {
    'tidak', 'bukan', 'jangan', 'ga', 'gak', 'gk', 'ngga', 'nggak',
    'enggak', 'nope', 'no', 'never', 'belum', 'tanpa'
}

# ============================================================================
# 3. TOXIC BIGRAMS (frasa 2 kata yang toxic)
# ============================================================================

TOXIC_BIGRAMS = {
    # Kombinasi penghinaan
    'bodoh banget': 5, 'tolol banget': 5, 'goblok banget': 5,
    'sampah masyarakat': 6, 'tidak berguna': 4, 'ga berguna': 4,
    'mending mati': 6, 'mati aja': 6,
    
    # Gaming toxic
    'noob banget': 4, 'cupu banget': 4, 'player sampah': 5,
    'toxic player': 4, 'dog player': 5, 'trash player': 5,
    'ga ngerti': 3, 'sok jago': 3, 'jago kandang': 3,
    
    # Pattern cyberbullying
    'tolol afk': 5, 'push lah': 2, 'def lah': 2,
    'otak udang': 5, 'otak atik': 5, 'otak dongo': 5,
}

# Gabungkan semua lexicon
ALL_LEXICONS_V2 = {
    'extreme': extreme_toxic_v2,
    'personal': personal_attack_v2,
    'gaming': gaming_toxic_v2,
    'negative': negative_words_v2
}

print(f"\n✅ Lexicon V2 Loaded:")
print(f"   Extreme Toxic    : {len(extreme_toxic_v2)} words")
print(f"   Personal Attack  : {len(personal_attack_v2)} words")
print(f"   Gaming Toxic     : {len(gaming_toxic_v2)} words")
print(f"   Negative Words   : {len(negative_words_v2)} words")
print(f"   Toxic Bigrams    : {len(TOXIC_BIGRAMS)} phrases")
print(f"   Total Coverage   : {sum(len(lex) for lex in ALL_LEXICONS_V2.values()) + len(TOXIC_BIGRAMS)} entries")

# ============================================================================
# 4. HELPER FUNCTIONS
# ============================================================================

def generate_bigrams(text):
    """Generate bigrams dari text"""
    words = text.split()
    bigrams = []
    for i in range(len(words) - 1):
        bigrams.append(f"{words[i]} {words[i+1]}")
    return bigrams


def check_negation(words, idx):
    """
    Cek apakah kata pada index idx didahului oleh negasi
    
    Returns:
        True jika ada negasi dalam jarak 1-2 kata sebelumnya
    """
    # Cek 1-2 kata sebelumnya
    for offset in [1, 2]:
        if idx >= offset:
            if words[idx - offset] in NEGATION_WORDS:
                return True
    return False


def normalize_leet_speak(word):
    """Normalize l33t speak ke bentuk normal"""
    leet_map = {
        '0': 'o', '1': 'i', '3': 'e', '4': 'a', 
        '5': 's', '7': 't', '8': 'b', '9': 'g'
    }
    normalized = word
    for leet, normal in leet_map.items():
        normalized = normalized.replace(leet, normal)
    return normalized


# ============================================================================
# 5. IMPROVED SCORING FUNCTION
# ============================================================================

def calculate_toxicity_score_v2(text, original_text):
    """
    Fungsi scoring yang ditingkatkan dengan:
    - Negation handling
    - Bigram detection  
    - L33t speak normalization
    - Better context awareness
    
    Returns:
        score (int): Total toxicity score
        matched_items (list): List of (word/phrase, category, score, negated)
        severity (str): Severity level
    """
    score = 0
    matched_items = []
    words = text.split()
    
    # ========================================
    # STEP 1: Check BIGRAMS first (higher priority)
    # ========================================
    bigrams = generate_bigrams(text)
    for bigram in bigrams:
        if bigram in TOXIC_BIGRAMS:
            bigram_score = TOXIC_BIGRAMS[bigram]
            score += bigram_score
            matched_items.append((bigram, 'bigram', bigram_score, False))
    
    # ========================================
    # STEP 2: Check UNIGRAMS with negation
    # ========================================
    for idx, word in enumerate(words):
        # Normalize l33t speak
        normalized_word = normalize_leet_speak(word)
        
        is_negated = check_negation(words, idx)
        word_to_check = normalized_word
        
        # Check in each lexicon category
        matched = False
        
        # Check extreme toxic
        if word_to_check in extreme_toxic_v2:
            word_score = extreme_toxic_v2[word_to_check]
            if is_negated:
                word_score = 0  # Negate completely for extreme words
            score += word_score
            matched_items.append((word, 'extreme', word_score, is_negated))
            matched = True
        
        # Check personal attack
        elif word_to_check in personal_attack_v2:
            word_score = personal_attack_v2[word_to_check]
            if is_negated:
                word_score = max(0, word_score - 3)  # Reduce significantly
            score += word_score
            matched_items.append((word, 'personal', word_score, is_negated))
            matched = True
        
        # Check gaming toxic
        elif word_to_check in gaming_toxic_v2:
            word_score = gaming_toxic_v2[word_to_check]
            if is_negated:
                word_score = max(0, word_score - 2)  # Reduce moderately
            score += word_score
            matched_items.append((word, 'gaming', word_score, is_negated))
            matched = True
        
        # Check negative words
        elif word_to_check in negative_words_v2:
            word_score = negative_words_v2[word_to_check]
            if is_negated:
                word_score = 0  # Complete negation for negative words
            score += word_score
            matched_items.append((word, 'negative', word_score, is_negated))
            matched = True
    
    # ========================================
    # STEP 3: Repetition bonus
    # ========================================
    word_counts = Counter(words)
    for word, count in word_counts.items():
        if count >= 3:  # Kata diulang 3x atau lebih
            # Cek apakah kata toxic
            norm_word = normalize_leet_speak(word)
            if any(norm_word in lex for lex in ALL_LEXICONS_V2.values()):
                repetition_bonus = min(count - 2, 3)  # Max bonus 3
                score += repetition_bonus
                matched_items.append((f"[REPEAT:{word}]", 'repetition', repetition_bonus, False))
    
    # ========================================
    # STEP 4: Determine severity
    # ========================================
    if score == 0:
        severity = 'safe'
    elif score <= 2:
        severity = 'minor_negative'
    elif score <= 5:
        severity = 'moderate_toxic'
    elif score <= 8:
        severity = 'severe_toxic'
    else:
        severity = 'extreme_cyberbullying'
    
    return score, matched_items, severity


def categorize_comment_v2(score, severity, matched_items):
    """
    Kategorisasi dengan logic yang ditingkatkan
    """
    # Cek apakah ada serious toxicity (extreme/personal/bigram)
    has_serious = any(
        item[1] in ['extreme', 'personal', 'bigram'] 
        for item in matched_items
    )
    
    # Decision logic
    if score >= 4 and has_serious:
        return 'cyberbullying'
    elif score >= 8:  # Very high score regardless
        return 'cyberbullying'
    else:
        return 'non-cyberbullying'


print("\n" + "=" * 80)
print("✅ Improved Detection Functions Ready!")
print("=" * 80)
print("\n📌 Key Improvements:")
print("   ✓ Negation handling (tidak, bukan, jangan, etc.)")
print("   ✓ Bigram detection (toxic phrases)")
print("   ✓ L33t speak normalization (b0d0h → bodoh)")
print("   ✓ Repetition detection")
print("   ✓ Context-aware scoring")
print("\n" + "=" * 80)
````

````python
# --- CELL 6: ADVANCED CYBERBULLYING DETECTION WITH SCORING SYSTEM ---
print("=" * 80)
print("--- Memulai Proses Advanced Cyberbullying Categorization ---")
print("=" * 80)

# ============================================================================
# LEXICON DENGAN KATEGORI DAN BOBOT
# ============================================================================

# Kategori 1: Kata Kasar Ekstrem & Ancaman (BOBOT TINGGI = 5)
# Ini adalah cyberbullying yang jelas
extreme_toxic = {
    # Ancaman & perintah menyakiti diri
    'mati': 5, 'bunuh diri': 5, 'pergi mati': 5, 'die': 5, 'drop dead': 5,
    'mati aja': 5, 'mati kamu': 5, 'bunuh': 5,
    
    # Kata kasar seksual
    'kontol': 5, 'kntl': 5, 'kintil': 5, 'memek': 5, 'mek': 5, 'pepek': 5,
    'pep3k': 5, 'pepeq': 5, 'peler': 5, 'titit': 5, 'titid': 5, 'tititmu': 5,
    
    # Kata kasar ekstrem
    'bajingan': 5, 'bangsat': 5, 'bangke': 5, 'brengsek': 5,
    'fuck': 5, 'fck': 5, 'fcking': 5, 'fucking': 5, 'fk u': 5,
    'bitch': 5, 'bicth': 5, 'btch': 5, 'asshole': 5, 'ashole': 5,
    'tai anjing': 5, 'tai babi': 5,
}

# Kategori 2: Penghinaan Personal & Diskriminatif (BOBOT SEDANG-TINGGI = 4)
# Kata yang menarget individu secara personal
personal_attack = {
    # Penghinaan intelektual
    'bodoh': 4, 'tolol': 4, 'bego': 4, 'idiot': 4, 'dungu': 4, 'goblok': 4,
    'bodo': 4, 'bodoamat': 4, 'tolil': 4, 'tololll': 4, 'begok': 4, 'begoo': 4,
    'gblk': 4, 'gblg': 4, 'stupid': 4, 'dumb': 4, 'dumbass': 4, 'retard': 4,
    'moron': 4, 'otak udang': 4, 'otak atik': 4, 'otak dongo': 4,
    
    # Penghinaan fisik/penampilan
    'jelek': 4, 'burik': 4, 'buruk': 4, 'sampah': 4, 'ampas': 4,
    
    # Hewan sebagai penghinaan
    'monyet': 4, 'anjing': 4, 'babi': 4, 'asu': 4, 'kampret': 4,
    'kodok': 4, 'cacing': 4, 'kecebong': 4, 'dog': 4,
    
    # Variasi anjing
    'anjg': 4, 'anjim': 4, 'njing': 4, 'anying': 4, 'anjink':4,
    
    # Diskriminatif
    'autis': 4, 'hina': 4, 'najis': 4,
}

# Kategori 3: Toxic Gaming Spesifik (BOBOT SEDANG = 3)
# Kata yang toxic dalam konteks gaming, tapi belum tentu cyberbullying
gaming_toxic = {
    # Skill-shaming
    'noob': 3, 'nob': 3, 'nub': 3, 'newbie': 3, 'cupu': 3, 'cupuu': 3,
    'lemah': 3, 'lemot': 3, 'lemottt': 3, 'useless': 3,
    
    # Perilaku negatif game
    'feeder': 3, 'feder': 3, 'fdr': 3, 'ngefeed': 3, 'inting': 3,
    'afk': 3, 'afkers': 3, 'beacon': 3, 'beakun': 3,
    
    # Toxic player behavior
    'toxic': 3, 'toxik': 3, 'toksek': 3, 'toxic player': 3,
    'player sampah': 3, 'ga guna': 3, 'tidak guna': 3,
    'ga berguna': 3, 'tidak berguna': 3, 'ga ngerti main': 3,
    
    # Gaming slang
    'sok jago': 3, 'jago kandang': 3, 'alay': 3,
    'loser': 3, 'ez game': 3, 'ez clap': 3, 'dogshit': 3, 'dog player': 3,
}

# Kategori 4: Kata Negatif Biasa (BOBOT RENDAH = 1-2)
# Kata negatif tapi bukan cyberbullying, lebih ke kritik atau keluhan
negative_words = {
    # Kritik umum
    'parah': 1, 'buruk': 1, 'jelek': 1,
    
    # Ekspresi negatif ringan
    'bacot': 2, 'bacod': 2, 'bacott': 2, 'cocot': 2, 'cocod': 2,
    'tai': 2, 'taii': 2, 'tayi': 2,
    'shit': 2, 'sh1t': 2, 'shet': 2, 'sheet': 2,
    
    # Slang negatif
    'anjir': 1, 'anjirr': 1, 'wtf': 2,
    'darat': 1,  # Konteks spesifik MLBB, tidak selalu toxic
}

# Kategori 5: Pola Cyberbullying (pattern-based)
# Frasa yang mengindikasikan cyberbullying lebih kuat
cyberbullying_patterns = [
    'tolol afk', 'push lah goblok', 'def lah tolol',
    'intentionally feed', 'wtf noob', 'lmao loser',
    'ga ngerti main', 'sok jago', 'jago kandang',
    'player sampah', 'toxic player', 'otak udang',
    'otak atik', 'otak dongo', 'tai anjing', 'tai babi',
    'bunuh diri', 'pergi mati', 'drop dead', 'fk u',
]

# ============================================================================
# FUNGSI SCORING & CATEGORIZATION
# ============================================================================

def calculate_cyberbullying_score(text, original_text):
    """
    Menghitung skor cyberbullying dari teks.
    
    Returns:
        score (int): Total skor
        matched_words (list): Kata-kata yang cocok dengan kategori
        severity (str): Tingkat keparahan
    """
    score = 0
    matched_words = []
    
    # 1. Cek kata dari berbagai kategori
    words_in_text = text.split()
    
    for word in words_in_text:
        # Cek extreme toxic
        if word in extreme_toxic:
            score += extreme_toxic[word]
            matched_words.append((word, 'extreme', extreme_toxic[word]))
        
        # Cek personal attack
        elif word in personal_attack:
            score += personal_attack[word]
            matched_words.append((word, 'personal_attack', personal_attack[word]))
        
        # Cek gaming toxic
        elif word in gaming_toxic:
            score += gaming_toxic[word]
            matched_words.append((word, 'gaming_toxic', gaming_toxic[word]))
        
        # Cek negative words
        elif word in negative_words:
            score += negative_words[word]
            matched_words.append((word, 'negative', negative_words[word]))
    
    # 2. Cek pola cyberbullying (menggunakan original text yang sudah di-clean)
    pattern_bonus = 0
    for pattern in cyberbullying_patterns:
        if pattern in text or pattern in original_text.lower():
            pattern_bonus += 3
            matched_words.append((pattern, 'pattern', 3))
    
    score += pattern_bonus
    
    # 3. Bonus untuk akumulasi kata toxic (semakin banyak kata toxic, semakin tinggi skor)
    toxic_word_count = len([w for w in matched_words if w[1] in ['extreme', 'personal_attack']])
    if toxic_word_count >= 3:
        score += 2  # Bonus untuk akumulasi
    
    # 4. Tentukan severity
    if score == 0:
        severity = 'safe'
    elif score <= 2:
        severity = 'minor_negative'
    elif score <= 5:
        severity = 'moderate_toxic'
    elif score <= 8:
        severity = 'severe_toxic'
    else:
        severity = 'extreme_cyberbullying'
    
    return score, matched_words, severity


def categorize_comment(score, severity, matched_words):
    """
    Kategorisasi komentar berdasarkan score dan severity.
    
    Aturan:
    - Cyberbullying: score >= 4 DAN ada kata dari extreme/personal_attack
    - Non-cyberbullying: lainnya
    """
    
    # Cek apakah ada kata extreme atau personal attack
    has_serious_toxicity = any(
        w[1] in ['extreme', 'personal_attack', 'pattern'] 
        for w in matched_words
    )
    
    # Decision logic
    if score >= 4 and has_serious_toxicity:
        return 'cyberbullying'
    elif score >= 8:  # Score sangat tinggi, pasti cyberbullying
        return 'cyberbullying'
    else:
        return 'non-cyberbullying'


# ============================================================================
# PROSES KATEGORISASI
# ============================================================================

# Struktur: (username, original, cleaned, normalized, stopword_removed, stemmed, 
#            score, severity, category, matched_words)
comments_categorized = []

if 'comments_stemmed' in locals():
    print("\n📊 Processing stemmed comments dengan Advanced Scoring System...")
    
    for username, original, cleaned, normalized, stopword_removed, stemmed in comments_stemmed:
        # Hitung score dari stemmed text dan original
        score, matched_words, severity = calculate_cyberbullying_score(stemmed, original)
        
        # Kategorisasi
        category = categorize_comment(score, severity, matched_words)
        
        # Simpan hasil
        comments_categorized.append((
            username, original, cleaned, normalized, 
            stopword_removed, stemmed, score, severity, 
            category, matched_words
        ))
    
    # ============================================================================
    # TAMPILKAN HASIL
    # ============================================================================
    
    print("\n" + "=" * 80)
    print("📋 HASIL CATEGORIZATION (20 Data Pertama)")
    print("=" * 80)
    
    if comments_categorized:
        for idx, (username, original, cleaned, normalized, stopword_removed, 
                  stemmed, score, severity, category, matched_words) in enumerate(comments_categorized[:20], 1):
            
            print(f"\n【{idx}】 {username}")
            print(f"Original    : {original}")
            print(f"Processed   : {stemmed}")
            print(f"Score       : {score}")
            print(f"Severity    : {severity}")
            print(f"Category    : {category.upper()}")
            
            if matched_words:
                print(f"Matched     : {', '.join([f'{w[0]}({w[2]})' for w in matched_words[:5]])}")
            
            print("-" * 80)
    else:
        print("❌ Tidak ada data yang dikategorisasi.")
    
    # ============================================================================
    # STATISTIK HASIL
    # ============================================================================
    
    print("\n" + "=" * 80)
    print("📈 STATISTIK HASIL KATEGORISASI")
    print("=" * 80)
    
    # Hitung per kategori
    cyberbullying_count = sum(1 for item in comments_categorized if item[8] == 'cyberbullying')
    non_cyberbullying_count = len(comments_categorized) - cyberbullying_count
    
    # Hitung per severity
    severity_counts = {}
    for item in comments_categorized:
        sev = item[7]
        severity_counts[sev] = severity_counts.get(sev, 0) + 1
    
    # Hitung score statistics
    scores = [item[6] for item in comments_categorized]
    avg_score = sum(scores) / len(scores) if scores else 0
    max_score = max(scores) if scores else 0
    min_score = min(scores) if scores else 0
    
    print(f"\n🔢 KATEGORI UTAMA:")
    print(f"  Total Komentar        : {len(comments_categorized)}")
    print(f"  Cyberbullying         : {cyberbullying_count} ({cyberbullying_count/len(comments_categorized)*100:.1f}%)")
    print(f"  Non-Cyberbullying     : {non_cyberbullying_count} ({non_cyberbullying_count/len(comments_categorized)*100:.1f}%)")
    
    print(f"\n📊 SEVERITY BREAKDOWN:")
    for sev, count in sorted(severity_counts.items(), key=lambda x: x[1], reverse=True):
        percentage = count / len(comments_categorized) * 100
        print(f"  {sev:25s}: {count:4d} ({percentage:5.1f}%)")
    
    print(f"\n🎯 SCORE STATISTICS:")
    print(f"  Average Score         : {avg_score:.2f}")
    print(f"  Maximum Score         : {max_score}")
    print(f"  Minimum Score         : {min_score}")
    
    # Top toxic comments
    print(f"\n⚠️  TOP 5 MOST TOXIC COMMENTS:")
    top_toxic = sorted(comments_categorized, key=lambda x: x[6], reverse=True)[:5]
    for idx, (username, original, _, _, _, _, score, severity, category, matched) in enumerate(top_toxic, 1):
        print(f"  {idx}. Score {score} | {original[:60]}...")
    
    print("\n" + "=" * 80)
    
else:
    print("❌ Please run the stemming cell first.")

print("\n✅ Proses Advanced Lexicon-Based Categorization Selesai")
print("=" * 80)
````
