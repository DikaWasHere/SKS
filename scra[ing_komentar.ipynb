```python
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
import csv

# Ganti dengan API key kamu
API_KEY = 'AIzaSyCHIShRhATsUAQ1OxoxLPm5da5tB1nxNd0'
# Tambahkan 10 VIDEO_ID ke dalam daftar ini
VIDEO_IDS = [
    # 'cqzUa5E-nf8',
    # 'hDZPlWX8UkY',
    # 'OXr9HwyYqaI',
    # 'LpothaxPVEY',
    # 'yyt97LgXj7c',
    # '0m7G9lZAEUI',
    # 'nHuFpqp9YQw',
    # 'jaWZaX2t4AU',
    # 'bCCsL7MzHeg',
    '_LSTqgsQ-gI'
]

# Buat service YouTube
youtube = build('youtube', 'v3', developerKey=API_KEY)

def get_comments_for_video(video_id):
    """Gets comments for a single video ID."""
    comments = []
    next_page_token = None

    while True:
        try:
            request = youtube.commentThreads().list(
                part='snippet',
                videoId=video_id,
                maxResults=100,
                textFormat='plainText',
                pageToken=next_page_token
            )
            response = request.execute()

            for item in response['items']:
                snippet = item['snippet']['topLevelComment']['snippet']
                username = snippet['authorDisplayName']
                comment = snippet['textDisplay']
                comments.append((username, comment)) # Exclude video_id

            next_page_token = response.get('nextPageToken')

            if not next_page_token:
                break

        except HttpError as e:
            print(f"An HTTP error {e.resp.status} occurred for video {video_id}: {e.content}")
            # Handle specific error codes if needed, e.g., 404 for video not found
            if e.resp.status == 404:
                print(f"Video with ID {video_id} not found or is unavailable.")
            break # Exit the loop if an error occurs

    return comments


def get_comments_for_videos(video_ids):
    """Gets comments for a list of video IDs."""
    all_comments = []
    for video_id in video_ids:
        print(f"Getting comments for video: {video_id}")
        all_comments.extend(get_comments_for_video(video_id)) # Extend the list with comments from each video
    return all_comments


# Contoh penggunaan
all_video_comments = get_comments_for_videos(VIDEO_IDS)

```

# **Preprocessing**

```python
# --- CELL 1: CASE FOLDING ---
# Proses mengubah semua huruf menjadi huruf kecil (lowercase).

print("--- Memulai Proses Case Folding ---")

# Kita akan membuat list baru untuk menyimpan hasil
# Strukturnya: (username, original_comment, folded_comment)
comments_folded = []
for username, comment in all_video_comments:
    folded_comment = comment.lower()
    comments_folded.append((username, comment, folded_comment))

print("--- Proses Case Folding Selesai ---")
if 'comments_folded' in locals():
  print("--- First 5 Examples After Case Folding ---")
  for username, original, folded in comments_folded[:5]:
      print(f"Original: {original}")
      print(f"Folded  : {folded}\n---")
else:
  print("Please run the case folding cell first.")
```

```python
# --- CELL 2: CLEANING ---
# Proses membersihkan teks dari karakter-karakter yang tidak diperlukan.

import re

print("--- Memulai Proses Cleaning ---")

# Kita akan membuat list baru untuk menyimpan hasil
# Strukturnya: (username, original_comment, folded_comment, cleaned_comment)
comments_cleaned = []

for username, original_comment, folded_comment in comments_folded:
    # Remove URLs
    cleaned_comment = re.sub(r'http\S+', '', folded_comment)
    # Remove mentions (@usernames)
    cleaned_comment = re.sub(r'@\S+', '', cleaned_comment)
    # Remove hashtags (#hashtags)
    cleaned_comment = re.sub(r'#\S+', '', cleaned_comment)
    # Remove numbers
    cleaned_comment = re.sub(r'\d+', '', cleaned_comment)
    # Remove punctuation and special characters (keep only letters and spaces)
    cleaned_comment = re.sub(r'[^a-zA-Z\s]', '', cleaned_comment)
    # Remove extra whitespace
    cleaned_comment = re.sub(r'\s+', ' ', cleaned_comment).strip()

    comments_cleaned.append((username, original_comment, folded_comment, cleaned_comment))

print("--- Proses Cleaning Selesai ---")

# Displaying the first 5 results after cleaning for comparison
print("\n--- Hasil Cleaning (5 Data Pertama) ---")
if 'comments_cleaned' in locals():
  for username, original, folded, cleaned in comments_cleaned[:5]:
      print(f"Original: {original}")
      print(f"Folded  : {folded}")
      print(f"Cleaned : {cleaned}\n---")
else:
  print("Please run the cleaning cell first.")
```

```python
# --- CELL 3: NORMALISASI ---
print("--- Memulai Proses Normalisasi ---")

# 1. Kamus normalisasi sederhana (slang → baku)
kamus_normalisasi = {
    'aj': 'saja', 'aja': 'saja', 'ae': 'saja',
    'ama': 'sama',
    'ampe': 'sampai', 'sampe': 'sampai', 'smpe': 'sampai',
    'ane': 'saya', 'gua': 'saya', 'gue': 'saya', 'gw': 'saya',
    'anjg': 'anjing', 'anjim': 'anjing', 'anying': 'anjing',
    'aq': 'aku', 'q': 'aku',
    'bangt': 'banget', 'bener': 'benar', 'bgt': 'banget', 'bgd': 'banget', 'bnget': 'banget', 'bngt': 'banget', 'bat': 'banget',
    'blg': 'bilang',
    'blm': 'belum', 'blom': 'belum',
    'bkn': 'bukan',
    'brp': 'berapa',
    'bnyk': 'banyak', 'byk': 'banyak',
    'bs': 'bisa', 'bsa': 'bisa',
    'btr': 'bentar',
    'ciamik': 'bagus',
    'cm': 'cuma', 'cmn': 'cuma',
    'coach': 'pelatih',
    'coy': 'kawan', 'cuy': 'kawan', 'gan': 'kawan', 'bro': 'kawan',
    'd': 'di',
    'dah': 'sudah',
    'dalem': 'dalam', 'dlm': 'dalam',
    'dapet': 'dapat', 'dpt': 'dapat',
    'dgn': 'dengan',
    'dlu': 'dulu',
    'dr': 'dari', 'dri': 'dari',
    'drpd': 'daripada',
    'darat': 'darat',
    'emang': 'memang', 'emg': 'memang', 'emng': 'memang',
    'ente': 'kamu',
    'euy': '',
    'fix': 'pasti',
    'ga': 'tidak', 'gak': 'tidak', 'gk': 'tidak', 'ngga': 'tidak', 'nggak': 'tidak',
    'gaada': 'tidak ada', 'gakada': 'tidak ada', 'gada': 'tidak ada', 'gda': 'tidak ada',
    'gabisa': 'tidak bisa',
    'gajelas': 'tidak jelas',
    'gblk': 'goblok',
    'gelo': 'gila',
    'gg': 'hebat',
    'gimana': 'bagaimana', 'gmn': 'bagaimana',
    'gini': 'begini', 'gni': 'begini',
    'gitu': 'begitu', 'gt': 'begitu', 'gtu': 'begitu',
    'gpp': 'tidak apa-apa',
    'gatau': 'tidak tahu', 'gaktau': 'tidak tahu', 'gtau': 'tidak tahu',
    'hbs': 'habis',
    'ilang': 'hilang',
    'indo': 'indonesia',
    'jd': 'jadi', 'jdi': 'jadi',
    'jg': 'juga', 'jga': 'juga',
    'jgn': 'jangan',
    'jir': ' ',
    'anjir': ' ',
    'anjay': ' ',
    'kalo': 'kalau', 'kl': 'kalau', 'klo': 'kalau',
    'karna': 'karena', 'krn': 'karena',
    'kdg': 'kadang',
    'kek': 'seperti', 'kayak': 'seperti', 'ky': 'seperti', 'kyk': 'seperti',
    'knp': 'kenapa',
    'kntl': 'kontol',
    'kok': 'kenapa',
    'kpn': 'kapan',
    'lg': 'lagi', 'lgi': 'lagi',
    'liat': 'lihat',
    'lo': 'kamu', 'lu': 'kamu', 'elu': 'kamu',
    'mah': '',
    'makasi': 'terima kasih', 'makasih': 'terima kasih', 'mksih': 'terima kasih', 'tks': 'terima kasih', 'trims': 'terima kasih', 'thx': 'terima kasih',
    'min': 'admin',
    'mlh': 'malah',
    'mnit': 'menit',
    'moga': 'semoga', 'smoga': 'semoga',
    'msh': 'masih',
    'mulu': 'terus',
    'ngasih': 'memberi',
    'ngelawak': 'melawak',
    'ngerti': 'mengerti',
    'ngomong': 'berbicara',
    'nih': '',
    'njir': ' ',
    'nt': 'nanti', 'ntar': 'nanti', 'ntr': 'nanti',
    'org': 'orang', 'org2': 'orang-orang',
    'pake': 'pakai', 'pakek': 'pakai',
    'pdhl': 'padahal',
    'player': 'pemain',
    'pls': 'tolong', 'plis': 'tolong',
    'prnh': 'pernah',
    'sbnrnya': 'sebenarnya',
    'sdh': 'sudah',
    'sih': '',
    'skrg': 'sekarang',
    'sllu': 'selalu', 'slalu': 'selalu',
    'sm': 'sama',
    'sndiri': 'sendiri',
    'spt': 'seperti', 'sprti': 'seperti',
    'sy': 'saya', 'sya': 'saya',
    'tau': 'tahu',
    'tbtb': 'tiba-tiba',
    'tdk': 'tidak',
    'team': 'tim',
    'tpi': 'tapi', 'tp': 'tapi',
    'trs': 'terus', 'trus': 'terus',
    'tuh': '',
    'tutor': 'tutorial',
    'tetep': 'tetap',
    'udh': 'sudah',
    'utk': 'untuk',
    'wktu': 'waktu',
    'woi': '',
    'cok': '',
    'y': 'ya', 'yaa': 'ya', 'yah': 'ya',
    'yg': 'yang',
    'dn': 'dan'
}

# 2. Import SymSpell untuk spell correction
from symspellpy.symspellpy import SymSpell, Verbosity
import difflib

# Inisialisasi SymSpell
sym_spell = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)

# Tambahkan beberapa kata dasar Indonesia ke dictionary SymSpell
basic_vocab = list(kamus_normalisasi.values()) + [
    "saya", "tidak", "bisa", "benar", "sudah", "akan", "terima", "kasih",
    "banget", "orang", "kalau", "dengan", "sekarang", "terus", "nanti", "ya",
    "mengerti", "berbicara", "makan", "tidur", "main", "bola", "tim", "pemain"
]
for word in basic_vocab:
    sym_spell.create_dictionary_entry(word, 1)

# Fungsi spell correction dengan threshold similarity
def spell_correct(word, min_similarity=0.8):
    suggestions = sym_spell.lookup(word, Verbosity.CLOSEST, max_edit_distance=2)
    if not suggestions:
        return word
    suggestion = suggestions[0].term
    ratio = difflib.SequenceMatcher(None, word, suggestion).ratio()
    return suggestion if ratio >= min_similarity else word

# 3. Fungsi normalisasi hybrid
def normalize_text(text):
    words = text.split()
    normalized_words = []
    for w in words:
        lw = w.lower()
        if lw in kamus_normalisasi:  # jika ada di kamus slang
            normalized_words.append(kamus_normalisasi[lw])
        else:  # gunakan symspell
            normalized_words.append(spell_correct(lw))
    return " ".join(normalized_words)

# --- Gunakan hasil dari cell cleaning ---
comments_normalized = []
for username, original, folded, cleaned in comments_cleaned:
    normalized_comment = normalize_text(cleaned)
    comments_normalized.append((username, original, cleaned, normalized_comment))

# --- Tampilkan 5 hasil pertama ---
print("\n--- Hasil Normalisasi (5 Data Pertama) ---")
for _, original, cleaned, normalized in comments_normalized[:5]:
    print(f"Original: {original}")
    print(f"Cleaned : {cleaned}")
    print(f"After   : {normalized}\n---")
```
